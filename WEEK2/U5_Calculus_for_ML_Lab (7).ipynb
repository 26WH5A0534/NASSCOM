1. Derivatives — the slope of a function

1A. NUMERICAL DERIVATIVE (finite difference)

```python
def f(x):
    return x**2

h = 1e-6
slope_at_3 = (f(3 + h) - f(3)) / h
print('Numerical f\'(3):', round(slope_at_3, 4), ' (exact = 6)')

```

1B. SYMBOLIC DERIVATIVE (SymPy)

```python
expr = x**2
deriv = sp.diff(expr, x)
print('d/dx (x**2) =', deriv)


print('Symbolic f\'(3) =', deriv.subs(x, 3))
```

LAB EXERCISE 1 — Numerical vs symbolic derivative

```python
# 1. Numerical derivative at x = 2 (use h = 1e-6)

import numpy as np

def g(x):
    return x**3 + 2*x

h = 1e-6
x = 2
numerical_derivative = (g(x + h) - g(x)) / h

print('Numerical derivative:', numerical_derivative)
```

```python
# 2. Symbolic derivative of x**3 + 2*x

import sympy as sp

x = sp.symbols('x')
g_sym = x**3 + 2*x
derivative = sp.diff(g_sym, x)

print('Symbolic derivative:', derivative)
```

```python
# 3. Evaluate the symbolic derivative at x = 2 and compare

import sympy as sp

x = sp.symbols('x')
g_sym = x**3 + 2*x
derivative = sp.diff(g_sym, x)

h = 1e-6
numerical_derivative = (g_sym.subs(x, 2 + h) - g_sym.subs(x, 2)) / h
symbolic_value = derivative.subs(x, 2)

print('Symbolic derivative at x=2:', symbolic_value)
print('Numerical derivative at x=2:', numerical_derivative)
print('Difference:', abs(float(symbolic_value) - float(numerical_derivative)))
```

2. Partial derivatives & the gradient

2A. PARTIAL DERIVATIVES

```python
f2 = x**2 + 3*x*y + y**2
print('df/dx =', sp.diff(f2, x))
print('df/dy =', sp.diff(f2, y))
```

2B. THE GRADIENT (vector of partials)

```python
grad = [sp.diff(f2, v) for v in (x, y)]
print('grad f =', grad)


grad_at = [g.subs({x: 1, y: 2}) for g in grad]
print('grad f at (1, 2) =', grad_at)
```

LAB EXERCISE 2 — Compute a gradient

```python
# 1. dh/dx and dh/dy

import sympy as sp

x, y = sp.symbols('x y')

h2 = x**2 * y + sp.sin(y)

dh_dx = sp.diff(h2, x)
dh_dy = sp.diff(h2, y)

print('dh/dx:', dh_dx)
print('dh/dy:', dh_dy)
```

```python
# 2. Assemble the gradient list

gradient = [dh_dx, dh_dy]

print('Gradient:', gradient)
```

```python
# 3. Evaluate at (x=2, y=0)

gradient_at_point = [expr.subs({x: 2, y: 0}) for expr in gradient]

print('Gradient at (2, 0):', gradient_at_point)
```

3. The chain rule

3A. CHAIN RULE BY HAND vs SymPy

```python
by_hand = sp.cos(x**2) * 2*x
by_sympy = sp.diff(sp.sin(x**2), x)

print('By hand :', by_hand)
print('By SymPy:', by_sympy)
print('Match?  ', sp.simplify(by_hand - by_sympy) == 0)
```

3B. CHAINING THREE FUNCTIONS

```python
expr3 = (3*x + 1)**4
print('d/dx (3x+1)^4 =', sp.diff(expr3, x))   # 12*(3x+1)^3

```

4. Backpropagation — the chain rule in a network

 4A. FORWARD PASS

```python
X = np.random.randn(4, 3)
Y = np.random.randn(4, 1)
W1 = np.random.randn(3, 5) * 0.1
W2 = np.random.randn(5, 1) * 0.1

z1 = X @ W1
h  = np.maximum(0, z1)
y_hat = h @ W2
loss = ((y_hat - Y) ** 2).mean()
print('Initial loss:', round(loss, 4))
```

4B. BACKWARD PASS (gradients via the chain rule)

```python
dy   = 2 * (y_hat - Y) / Y.size
dW2  = h.T @ dy
dh   = dy @ W2.T
dz1  = dh * (z1 > 0)
dW1  = X.T @ dz1

print('dW1 shape:', dW1.shape, '(matches W1)')
print('dW2 shape:', dW2.shape, '(matches W2)')

```

 4C. ONE GRADIENT-DESCENT STEP SHOULD LOWER THE LOSS

```python

lr = 0.1
W1 -= lr * dW1
W2 -= lr * dW2


h_new = np.maximum(0, X @ W1)
loss_new = ((h_new @ W2 - Y) ** 2).mean()
print('Loss before:', round(loss, 4))
print('Loss after :', round(loss_new, 4), '-> should be lower')
```

LAB EXERCISE 4 — Derive the gradients for a 2-layer network

```python
# 1. Forward pass: z1, h = ReLU(z1), y_hat, loss

import numpy as np

Xb = np.random.randn(6, 4)
Yb = np.random.randn(6, 1)
Wa = np.random.randn(4, 8) * 0.1
Wb = np.random.randn(8, 1) * 0.1

z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb
loss = np.mean((y_hat - Yb)**2)

print('Loss:', loss)
```

```python
# 2. Backward pass: dy, dWb, dh, dz1, dWa

dy = 2 * (y_hat - Yb) / Yb.shape[0]
dWb = h.T @ dy
dh = dy @ Wb.T
dz1 = dh * (z1 > 0)
dWa = Xb.T @ dz1

print('dWb:\n', dWb)
print('dWa:\n', dWa)
```

```python
# 3. One step with lr = 0.05; print loss before and after

lr = 0.05

loss_before = loss

Wa -= lr * dWa
Wb -= lr * dWb

z1 = Xb @ Wa
h = np.maximum(0, z1)
y_hat = h @ Wb
loss_after = np.mean((y_hat - Yb)**2)

print('Loss before:', loss_before)
print('Loss after:', loss_after)
```

5. Hessian & gradient descent

5A. THE HESSIAN (matrix of second derivatives = curvature)

```python


import sympy as sp

x, y = sp.symbols('x y')

f5 = x**2 + 3*x*y + y**2
H = sp.hessian(f5, (x, y))

print('Hessian of f:')
sp.pprint(H)
```

5B. GRADIENT DESCENT ON A SIMPLE FUNCTION

```python
xv = 0.0
lr = 0.2
for step in range(15):
    grad = 2 * (xv - 4)
    xv = xv - lr * grad
print('Converged x:', round(xv, 3), ' (true minimum = 4)')
```

LAB EXERCISE 5 — Hessian + gradient descent

```python
# 1. Hessian of x**4 + y**2

import sympy as sp

x, y = sp.symbols('x y')

f = x**4 + y**2
H = sp.hessian(f, (x, y))

print('Hessian:')
sp.pprint(H)
```

```python
# 2. Gradient descent to minimise (x - 7)**2

xv = 0.0
lr = 0.1

for step in range(20):
    grad = 2 * (xv - 7)
    xv = xv - lr * grad

print('Final x:', xv)
```
